```markdown
# STEP 1: Import Required Libraries and Load Dataset Safely

In this step, we initialize our analytical environment, verify that the dataset exists, load the data into a pandas DataFrame, and print the first 5 rows to understand the structure.
```

````python
import os
import warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

warnings.filterwarnings('ignore')

# File path definition
dataset_path = 'nepal_flood_weather_dataset_kaggle_2023_2026.csv'

# Checking if the file exists safely
if os.path.exists(dataset_path):
    print(f"[✓] File found successfully at: {os.path.abspath(dataset_path)}")

    # Loading the dataset
    df = pd.read_csv(dataset_path)

    # Basic metadata extraction
    num_rows, num_cols = df.shape
    print(f"[✓] Dataset loaded successfully!")
    print(f"-> Total Rows (Records): {num_rows}")
    print(f"-> Total Columns (Features): {num_cols}\n")

    # Display first 5 rows
    print("=== FIRST 5 ROWS OF THE DATASET ===")
    display(df.head())
else:
    print(f"[X] ERROR: Dataset not found at '{dataset_path}'. Please upload the CSV first!")
````

```markdown
# STEP 2: Basic Dataset Inspection

In this step, we perform a deep inspection of the dataset structure. We look at the last 5 records, verify column names and data types, examine structural summaries using `info()`, analyze numeric properties, and check text columns.
```

````python
# 1. Print last 5 rows
print("=== LAST 5 ROWS OF THE DATASET ===")
display(df.tail())

# 2. Print shape and column information
print("\n=== DATASET SHAPE ===")
print(f"Rows: {df.shape[0]}, Columns: {df.shape[1]}")

# 3. Print DataFrame detailed structural info
print("\n=== DATAFRAME INFO ===")
df.info()

# 4. Describe numerical columns
print("\n=== STATISTICAL SUMMARY (NUMERICAL COLUMNS) ===")
display(df.describe())

# 5. Describe categorical columns
print("\n=== STATISTICAL SUMMARY (CATEGORICAL COLUMNS) ===")
display(df.describe(include='object'))

# 6. Unique values count for each column
print("\n=== UNIQUE VALUES COUNT PER COLUMN ===")
for col in df.columns:
    print(f"{col}: {df[col].nunique()} unique values")
````

```markdown
# STEP 3: Checking for Missing Values

Here we analyze the dataset for completeness, calculating the sum and percentage of missing values (NaN) for each feature and plotting a horizontal bar chart of the missing data.
```

````python
# 1. Calculate missing values and percentages
missing_count = df.isnull().sum()
missing_percentage = (df.isnull().sum() / len(df)) * 100

# Combine into a single readable DataFrame
missing_df = pd.DataFrame({
    'Missing Count': missing_count,
    'Percentage (%)': missing_percentage
})

print("=== MISSING VALUES SUMMARY ===")
display(missing_df.sort_values(by='Missing Count', ascending=False))

# 2. Plotting a horizontal bar chart of missing values
plt.figure(figsize=(12, 6))
sns.barplot(
    x=missing_percentage,
    y=missing_percentage.index,
    palette="viridis",
    edgecolor="black"
)

plt.title("Percentage of Missing Values per Column", fontsize=14, pad=15)
plt.xlabel("Percentage Missing (%)", fontsize=12)
plt.ylabel("Columns", fontsize=12)
plt.xlim(0, 100) # Setting scale from 0 to 100% to visualize data fullness
plt.grid(axis='x', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
````

```markdown
# STEP 4: Check and Remove Duplicates

In this step, we identify duplicate rows inside the dataset. If any duplicates are found, they are removed from a fresh, cleaned copy of our DataFrame to preserve data integrity.
```

````python
# 1. Count duplicate rows
duplicate_count = df.duplicated().sum()
print("=== DUPLICATE ANALYSIS ===")
print(f"Total Duplicate Rows Found: {duplicate_count}")

# 2. Show sample duplicate rows if any exist
if duplicate_count > 0:
    print("\nShowing sample duplicate rows:")
    display(df[df.duplicated()].head())
else:
    print("\nNo duplicate rows detected in the dataset!")

# 3. Create a clean copy and remove duplicates safely
df_cleaned = df.copy()
if duplicate_count > 0:
    df_cleaned = df_cleaned.drop_duplicates()
    print(f"\nDuplicates removed safely!")

print(f"Rows in original dataframe: {df.shape[0]}")
print(f"Rows in cleaned dataframe (without duplicates): {df_cleaned.shape[0]}")
````

```markdown
# STEP 5: Understand Columns (Classification & Keyword Matching)

In this step, we programmatically classify our dataset's features into numerical, categorical, and datetime formats. We also dynamically search for high-priority flood and weather-related features.
```

````python
# 1. Automatic classification based on data types
numeric_cols = df_cleaned.select_dtypes(include=[np.number]).columns.tolist()
categorical_cols = df_cleaned.select_dtypes(include=['object', 'category']).columns.tolist()

# Identify possible datetime columns (by name or content)
datetime_cols = [col for col in df_cleaned.columns if 'date' in col.lower() or 'time' in col.lower()]

# Adjust lists so date column is not duplicated
for d_col in datetime_cols:
    if d_col in categorical_cols:
        categorical_cols.remove(d_col)

print("=== COLUMN CLASSIFICATION ===")
print(f"Numerical Columns ({len(numeric_cols)}): {numeric_cols}")
print(f"Categorical Columns ({len(categorical_cols)}): {categorical_cols}")
print(f"Datetime Column Candidates ({len(datetime_cols)}): {datetime_cols}")

# 2. Search for important flood/weather hazard-related columns using keywords
keywords = ['flood', 'rainfall', 'rain', 'water', 'river', 'level', 'disaster', 'risk', 'severity', 'damage', 'warning', 'discharge']
matched_cols = []

for col in df_cleaned.columns:
    for kw in keywords:
        if kw in col.lower():
            matched_cols.append(col)
            break

print("\n=== FLOOD & WEATHER KEYWORD MATCHED COLUMNS ===")
print(f"Detected Target Columns: {matched_cols}")
````

```markdown
# STEP 6: Clean Data Safely (Datetime Conversion & Space Stripping)

In this step, we ensure proper formatting of date, strip categorical spaces, and check for infinite values to prepare the data for analysis.
```

````python
# 1. Parse date column to pandas Datetime format safely
print("=== PARSING DATETIME ===")
print(f"Original 'date' type: {df_cleaned['date'].dtype}")
df_cleaned['date'] = pd.to_datetime(df_cleaned['date'], format='%d-%m-%Y', errors='coerce')
print(f"Updated 'date' type: {df_cleaned['date'].dtype}")

# 2. Strip whitespaces from all categorical text columns
print("\n=== STRIPPING CATEGORICAL WHITESPACES ===")
for col in ['location', 'river', 'basin']:
    original_uniques = df_cleaned[col].nunique()
    df_cleaned[col] = df_cleaned[col].astype(str).str.strip()
    new_uniques = df_cleaned[col].nunique()
    print(f"Column '{col}': Unique values changed from {original_uniques} to {new_uniques}")

# 3. Check for infinite values in numerical columns
print("\n=== INFINITE VALUE CHECK ===")
inf_cols = {}
for col in numeric_cols:
    inf_count = np.isinf(df_cleaned[col]).sum()
    if inf_count > 0:
        inf_cols[col] = inf_count

if inf_cols:
    print(f"Infinite values found: {inf_cols}")
else:
    print("No infinite values detected in numerical columns!")

# Display basic metadata again to verify
print("\n=== CLEANED DATA SUMMARY ===")
print(df_cleaned.info())
````

```markdown
# STEP 7: Check Data Quality & Physical Constraints

Here, we validate if the numerical records satisfy real-world meteorological logic and physical constraints (e.g. humidity bounds, precipitation consistency, and wind speeds).
```

````python
# 1. Check relative humidity bounds
humidity_violations = df_cleaned[df_cleaned['relative_humidity_mean_pct'] > 100].shape[0]

# 2. Check if rain exceeds total precipitation
rain_precip_violations = df_cleaned[df_cleaned['rain_mm'] > df_cleaned['precipitation_mm']].shape[0]

# 3. Check if sustained wind speed exceeds maximum wind gust
gust_speed_violations = df_cleaned[df_cleaned['wind_speed_max_kmh'] > df_cleaned['wind_gusts_max_kmh']].shape[0]

# 4. Check if wind direction degrees are within physical compass constraints [0, 360]
wind_dir_violations = df_cleaned[(df_cleaned['wind_direction_dominant_deg'] < 0) | (df_cleaned['wind_direction_dominant_deg'] > 360)].shape[0]

print("=== PHYSICAL METEOROLOGICAL CONSTRAINTS REPORT ===")
print(f"-> Relative Humidity > 100% Violations  : {humidity_violations} records")
print(f"-> Rain exceeds Precipitation Violations : {rain_precip_violations} records")
print(f"-> Wind Speed > Wind Gust Violations    : {gust_speed_violations} records")
print(f"-> Wind Direction out of range [0, 360] : {wind_dir_violations} records")
````

```markdown
# STEP 8: Resolve Meteorological and Physical Anomalies

We dynamically correct records where rain exceeds precipitation by setting precipitation equal to the rain value. We also inspect the known high-elevation anomaly at Kusum station and review statistical distributions.
```

````python
# 1. Calibrate Precipitation where Rain exceeds Precipitation
print("=== BEFORE CALIBRATION ===")
violating_mask = df_cleaned['rain_mm'] > df_cleaned['precipitation_mm']
print(f"Number of violating records: {violating_mask.sum()}")

# Adjust precipitation to equal rain where rain is higher
df_cleaned.loc[violating_mask, 'precipitation_mm'] = df_cleaned.loc[violating_mask, 'rain_mm']

print("\n=== AFTER CALIBRATION ===")
post_violations = df_cleaned[df_cleaned['rain_mm'] > df_cleaned['precipitation_mm']].shape[0]
print(f"Number of violating records: {post_violations}")

# 2. Inspect elevation values per station (Kusum is high-elevation outlier)
print("\n=== STATION ELEVATIONS CHECK ===")
elevations_by_loc = df_cleaned.groupby('location')['elevation_m'].unique()
for loc, elev in elevations_by_loc.items():
    print(f"Location: {loc:<25} | Elevation: {elev[0]} meters")

# 3. Quick check of 2023 Temperature anomaly (if values are swapped with soil moisture or corrupted)
df_2023 = df_cleaned[df_cleaned['date'].dt.year == 2023]
print("\n=== 2023 TEMPERATURE STATS ===")
print(df_2023['temperature_mean_c'].describe())
````

```markdown
# STEP 9: Reconstruct and Fix 2023 Swapped Columns

There is a systematic dataset swap in the year 2023 where the mean temperature and soil moisture columns were interchanged. We dynamically swap them back to reconstruct proper logical values for 2023.
```

````python
# 1. Identify rows belonging to 2023
mask_2023 = df_cleaned['date'].dt.year == 2023

print("=== BEFORE COLUMN SWAP REPAIR (2023) ===")
print("Temperature Stats in 2023:")
print(df_cleaned.loc[mask_2023, 'temperature_mean_c'].describe())
print("\nSoil Moisture Stats in 2023:")
print(df_cleaned.loc[mask_2023, 'soil_moisture_0_100cm_m3m3'].describe())

# 2. Perform the safe swap using a temporary holder variable
temp_holder = df_cleaned.loc[mask_2023, 'temperature_mean_c'].copy()
df_cleaned.loc[mask_2023, 'temperature_mean_c'] = df_cleaned.loc[mask_2023, 'soil_moisture_0_100cm_m3m3']
df_cleaned.loc[mask_2023, 'soil_moisture_0_100cm_m3m3'] = temp_holder

print("\n=== AFTER COLUMN SWAP REPAIR (2023) ===")
print("Corrected Temperature Stats in 2023:")
print(df_cleaned.loc[mask_2023, 'temperature_mean_c'].describe())
print("\nCorrected Soil Moisture Stats in 2023:")
print(df_cleaned.loc[mask_2023, 'soil_moisture_0_100cm_m3m3'].describe())
````

```markdown
# STEP 10: Handle Unphysical Extremes & Subset Filtering

Following our column swap, we observe that the corrected 2023 temperature features still contain unphysical spikes (like 268.3°C) and zeros. In this step, we identify these rows and create a clean filtered subset 'df_clean' for standard physical thresholds.
```

````python
# 1. Count records with unphysical values post-repair
unphysical_temp = df_cleaned[df_cleaned['temperature_mean_c'] > 60].shape[0]
unphysical_sm = df_cleaned[(df_cleaned['soil_moisture_0_100cm_m3m3'] < 0) | (df_cleaned['soil_moisture_0_100cm_m3m3'] > 1)].shape[0]

print("=== POST-REPAIR UNPHYSICAL RECORDS ===")
print(f"-> Records with Temperature > 60°C      : {unphysical_temp}")
print(f"-> Records with Soil Moisture outside [0, 1] : {unphysical_sm}")

# 2. Inspecting a sample of these weird records
if unphysical_temp > 0:
    print("\nSample of extreme temperature rows:")
    display(df_cleaned[df_cleaned['temperature_mean_c'] > 60].head())

# 3. Create a strict, physically consistent analytical subset
# We filter out records with unphysical temperature and keep standard soil moisture range
df_clean = df_cleaned[
    (df_cleaned['temperature_mean_c'] >= -10) &
    (df_cleaned['temperature_mean_c'] <= 55) &
    (df_cleaned['soil_moisture_0_100cm_m3m3'] >= 0.01) &
    (df_cleaned['soil_moisture_0_100cm_m3m3'] <= 0.95)
].copy()

print("\n=== DATASETS COMPARISON ===")
print(f"Original Cleaned Copy Shape : {df_cleaned.shape}")
print(f"Physically Corrected Subset Shape : {df_clean.shape}")
print(f"Discarded/Anomalous Records: {len(df_cleaned) - len(df_clean)}")
````

# STEP 11: Feature Engineering & Temporal Extraction

We enrich the clean subset with temporal columns (`year`, `month`, `season`) tailored to Nepal's meteorological seasons to unlock deep insights into seasonal flood cycles.

````python
# 1. Extract year and month from the parsed date column
df_clean['year'] = df_clean['date'].dt.year
df_clean['month'] = df_clean['date'].dt.month

# 2. Define a function to map months to Nepal's meteorological seasons
def get_nepal_season(month):
    if month in [12, 1, 2]:
        return 'Winter'
    elif month in [3, 4, 5]:
        return 'Pre-Monsoon'
    elif month in [6, 7, 8, 9]:
        return 'Monsoon'
    else:
        return 'Post-Monsoon'

# Apply the season mapping
df_clean['season'] = df_clean['month'].apply(get_nepal_season)

print("=== TEMPORAL FEATURE ENGINEERING ===")
print("Successfully created columns: 'year', 'month', and 'season'.\n")

# 3. Check distribution of records across seasons
print("Distribution of records by Season:")
display(df_clean['season'].value_value_counts() if hasattr(df_clean['season'], 'value_value_counts') else df_clean['season'].value_counts())
````

# STEP 12: Grouped Aggregations & Seasonal Summaries

In this step, we perform grouped aggregations on our clean subset to analyze the mean and maximum extremes of key meteorological indicators across different seasons.

````python
# 1. Group by season and aggregate mean and max for key features
seasonal_summary = df_clean.groupby('season').agg({
    'precipitation_mm': ['mean', 'max'],
    'river_discharge_m3s': ['mean', 'max'],
    'soil_moisture_0_100cm_m3m3': ['mean', 'max']
})

print("=== SEASONAL METEOROLOGICAL AGGREGATIONS ===")
display(seasonal_summary)

# 2. Group by basin to check average river discharge per basin area
basin_summary = df_clean.groupby('basin')['river_discharge_m3s'].agg(['mean', 'max', 'count'])
print("\n=== RIVER DISCHARGE BY RIVER BASIN ===")
display(basin_summary.sort_values(by='mean', ascending=False))
````

# STEP 13: Dual-Axis Monthly Seasonal Trend Visualization

We plot a dual-axis trend chart displaying monthly average precipitation alongside monthly average river discharge to visually analyze their lag-lead correlation.

````python
# 1. Compute monthly averages from the clean subset
monthly_data = df_clean.groupby('month').agg({
    'precipitation_mm': 'mean',
    'river_discharge_m3s': 'mean'
}).reset_index()

# 2. Map month numbers to English Month Abbreviations
month_names = {1:'Jan', 2:'Feb', 3:'Mar', 4:'Apr', 5:'May', 6:'Jun', 7:'Jul', 8:'Aug', 9:'Sep', 10:'Oct', 11:'Nov', 12:'Dec'}
monthly_data['month_name'] = monthly_data['month'].map(month_names)

# 3. Create the dual-axis plot
fig, ax1 = plt.subplots(figsize=(12, 6))

sns.set_style("whitegrid")

# Plot precipitation as bars on ax1
color_precip = '#1f77b4'
ax1.set_xlabel('Month', fontsize=12, labelpad=10)
ax1.set_ylabel('Mean Precipitation (mm)', color=color_precip, fontsize=12)
bars = ax1.bar(monthly_data['month_name'], monthly_data['precipitation_mm'], color=color_precip, alpha=0.6, label='Precipitation (mm)')
ax1.tick_params(axis='y', labelcolor=color_precip)
ax1.set_ylim(0, monthly_data['precipitation_mm'].max() * 1.15)

# Instantiate a second axes that shares the same x-axis
ax2 = ax1.twinx()
color_discharge = '#d62728'
ax2.set_ylabel('Mean River Discharge (m³/s)', color=color_discharge, fontsize=12)
line = ax2.plot(monthly_data['month_name'], monthly_data['river_discharge_m3s'], color=color_discharge, linewidth=2.5, marker='o', label='River Discharge')
ax2.tick_params(axis='y', labelcolor=color_discharge)
ax2.set_ylim(0, monthly_data['river_discharge_m3s'].max() * 1.15)

plt.title('Nepal Monthly Weather-Hydrology Cycle: Precipitation vs. River Discharge', fontsize=14, pad=15, fontweight='bold')
fig.tight_layout()
plt.show()
````

### STEP 14: Univariate Numerical Feature Analysis

In this step, we analyze the distribution shape, central tendency, spread, and skewness of our main numerical variables (`river_discharge_m3s`, `precipitation_mm`, and `temperature_mean_c`) using KDE plots and boxplots.

````python
# Define the target features for univariate analysis
features_to_analyze = ['river_discharge_m3s', 'precipitation_mm', 'temperature_mean_c']

# Loop through each feature and plot its distribution and boxplot side-by-side
for col in features_to_analyze:
    fig, axes = plt.subplots(1, 2, figsize=(15, 5))

    # Calculate skewness statistic
    skew_val = df_clean[col].skew()

    # 1. KDE Distribution Plot
    sns.histplot(df_clean[col], kde=True, ax=axes[0], color='teal', bins=40, stat='density', alpha=0.5)
    axes[0].set_title(f'{col} - Distribution (Skewness: {skew_val:.2f})', fontsize=12, fontweight='bold')
    axes[0].set_xlabel(col, fontsize=10)
    axes[0].set_ylabel('Density', fontsize=10)
    axes[0].grid(True, linestyle='--', alpha=0.6)

    # 2. Boxplot
    sns.boxplot(x=df_clean[col], ax=axes[1], color='coral', flierprops=dict(marker='o', markerfacecolor='red', markersize=5, alpha=0.5))
    axes[1].set_title(f'{col} - Boxplot & Outliers', fontsize=12, fontweight='bold')
    axes[1].set_xlabel(col, fontsize=10)
    axes[1].grid(True, linestyle='--', alpha=0.6)

    plt.suptitle(f'Univariate Analysis: {col}', fontsize=14, fontweight='bold', y=1.02)
    plt.tight_layout()
    plt.show()

    # Print descriptive statistics
    print(f"=== Summary Statistics for {col} ===")
    print(df_clean[col].describe())
    print("-" * 80 + "\n")
````

### STEP 15: Categorical Columns Distribution Analysis

In this step, we analyze the distribution of records across our spatial categories: `basin`, `location`, and `river`. This will help us understand if the station sensor records are evenly distributed across regions and hydrologic catchments.

````python
# Set up the plotting grid for categorical features
cat_features = ['basin', 'location', 'river']
fig, axes = plt.subplots(3, 1, figsize=(12, 18))

# Loop through categorical features and plot counts
for i, col in enumerate(cat_features):
    # Get value counts in descending order
    order_counts = df_clean[col].value_counts()

    # Plot using countplot with custom order
    sns.countplot(
        data=df_clean,
        y=col,
        order=order_counts.index,
        ax=axes[i],
        palette='viridis',
        edgecolor='black'
    )

    # Add count labels to the bars
    for container in axes[i].containers:
        axes[i].bar_label(container, padding=5, fontsize=10, fontweight='semibold')

    axes[i].set_title(f'Record Count Distribution by {col.upper()}', fontsize=14, fontweight='bold', pad=10)
    axes[i].set_xlabel('Count (Number of Records)', fontsize=11)
    axes[i].set_ylabel(col.capitalize(), fontsize=11)
    axes[i].grid(axis='x', linestyle='--', alpha=0.5)

plt.tight_layout()
plt.show()

# Print absolute counts and percentages
for col in cat_features:
    print(f"=== Frequency Distribution for {col.upper()} ===")
    freq_df = pd.DataFrame({
        'Count': df_clean[col].value_counts(),
        'Percentage (%)': df_clean[col].value_counts(normalize=True) * 100
    })
    display(freq_df)
    print("-" * 80 + "\n")
````
